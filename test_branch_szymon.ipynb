```python
import numpy as np
from matplotlib import pyplot as plt

# Generowanie danych (np. wzrost 50 osób)
np.random.seed(42)
wzrost = np.random.normal(175, 8, 50)

# Podstawowe statystyki
print(f"Średnia: {np.mean(wzrost):.2f} cm")
print(f"Mediana: {np.median(wzrost):.2f} cm")
print(f"Odchylenie std: {np.std(wzrost):.2f} cm")

# Wykres pudełkowy (boxplot)
plt.boxplot(wzrost)
plt.title("Rozkład wzrostu")
plt.ylabel("Wzrost [cm]")
plt.show()
```
