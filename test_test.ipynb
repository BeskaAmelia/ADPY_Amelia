```python
import numpy as np
from matplotlib import pyplot as plt

# 1. Ustawienie ziarna i zapisanie wylosowanych liczb do zmiennej 'dane'
np.random.seed(123)
dane = np.random.normal(181, 7, 20)
print(dane)

# 2. Pierwszy histogram (domyślne przedziały)
plt.hist(dane)
plt.show()

# 3. Drugi histogram (ręcznie podane przedziały)
plt.hist(dane, bins=[158, 162, 166, 170, 174, 178, 182, 186, 190, 194, 198, 202])
plt.show()

# 4. Trzeci histogram (przedziały wygenerowane przez np.arange)
breaks = np.arange(155, 210, 5)
print(breaks)
plt.hist(dane, bins=breaks)
plt.show()
```
