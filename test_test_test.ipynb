```python

import numpy as np
import matplotlib.pyplot as plt

# Generowanie przykładowych danych (np. wyniki egzaminu)
np.random.seed(42)
wyniki = np.random.normal(70, 10, 1000)

# Podstawowe statystyki
print("Średnia:", np.mean(wyniki))
print("Mediana:", np.median(wyniki))
print("Odchylenie standardowe:", np.std(wyniki))
print("Minimum:", np.min(wyniki))
print("Maksimum:", np.max(wyniki))

# Histogram
plt.hist(wyniki, bins=20, edgecolor="black")

plt.title("Rozkład wyników egzaminu")
plt.xlabel("Wynik")
plt.ylabel("Liczba osób")
plt.grid(axis="y", alpha=0.3)

plt.show()
```
